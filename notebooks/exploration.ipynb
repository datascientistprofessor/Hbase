# GSOD 2025 Exploration

Quick ad-hoc look at a few raw station files before running the full
Spark pipeline. Point `DATA_DIR` at your local `data/raw` folder.

```python
import pandas as pd
import glob

DATA_DIR = "../data/raw"
files = glob.glob(f"{DATA_DIR}/*.csv")[:20]
print(f"Found {len(glob.glob(f'{DATA_DIR}/*.csv'))} station files total; previewing {len(files)}")

df = pd.concat((pd.read_csv(f) for f in files), ignore_index=True)
df.head()
```

```python
df.describe(include='all')
```
